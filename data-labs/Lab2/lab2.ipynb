## 1. Count property types
Load the file. Use value_counts() on property_type and show only the five most common types. What share of all listings is "Entire rental unit"? Use normalize=True, rounded to 3 decimals

```python
import pandas as pd

listings = pd.read_csv("data/listings.csv")
```

```python
# The share of "Entire rental unit" is 53%
listings["property_type"].value_counts(normalize=True).head(5).round(3)
```

## 2. The smallest neighbourhoods
Use value_counts() on neighbourhood_cleansed. Which 3 neighbourhoods have the fewest listings, and how many do they have?

```python
# Spånga-Tensta (89), Skärholmens (96) and Rinkeby-Tensta (108)
# have the fewest listings respectively
listings["neighbourhood_cleansed"].value_counts(ascending=True).head(3)
```

## 3. Average beds per room type
Calculate the average of beds for each room_type, rounded to two decimals. Then use agg(["count", "mean"]) on the same thing. Which average do you trust the least, and why?

```python
listings.groupby("room_type")["beds"].mean().round(2)
```

```python
# I trust the Hotel room the least since there are too few listings to get a reliable average
import numpy as np
listings.groupby("room_type")["beds"].agg(["count", "mean"]).round(2)
```
