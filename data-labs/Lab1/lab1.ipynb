## 7. Filter rows
(a) Make a mask for listings with two or more bedrooms. Use sum() and mean() on it: how many are there, and what share of all listings?

(b) Find the listings that are in Östermalms, are Entire home/apt and have two or more bedrooms. Start with "and" between the conditions and read the error. Then fix it with & and parentheses, and print the shape. Use and first, and read the error from the bottom up.

```python
# There are 1825 bedrooms
mask = listings["bedrooms"] >= 2
mask.sum()
```

```python
# About 40% of all listings have two or more bedrooms
mask.mean()
```

```python
# Each mask is a column so we need "&" so it can compare their rows
neighbourhood_cleansed = listings["neighbourhood_cleansed"] == "Östermalms"
room_type = listings["room_type"] == "Entire home/apt"
bedrooms = listings["bedrooms"] >= 2

listings[neighbourhood_cleansed and room_type and bedrooms]
```

```python
# & compares the masks row by row so this will fix it
listings[neighbourhood_cleansed & room_type & bedrooms].shape
```

## 8. Sort and pick
Find the five listings in Östermalms with the most reviews. Show name, room_type and number_of_reviews, with the biggest number of reviews first. Which room type does the listing at the top have?

```python
# The top most listing has "Private room" as room_type
mask = listings["neighbourhood_cleansed"] == "Östermalms"
listings[mask][["name", "room_type", "number_of_reviews"]].sort_values("number_of_reviews", ascending=False).head(5)
```
