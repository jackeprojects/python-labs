## 2. Clean the genre
How many different values does *genre* have? It should be 6 genres. Remove spaces and capital letters with `str.strip()` and `str.lower()` and count again. Look at the values that are left with `unique()` and fix the rest with `replace`. Finish with `str.title()`. How many genres do you have, and how many loans in each?

```python
# 28 genres before cleanup
loans["genre"].unique()
```

```python
# 9 genres after first cleanup
loans["genre"] = loans["genre"].str.strip().str.lower()
loans["genre"].unique()
```

```python
# 6 genres after second cleanup
loans["genre"] = loans["genre"].replace({"children's": "children", "science fiction" : "sci-fi", "sci fi" : "sci-fi"}).str.title()
loans["genre"].unique()
```

```python
# 163 in Crime
# 135 in Children
# 90 in Fantasy
# 90 in Sci-Fi
# 86 in Romance
# 59 in History
loans["genre"].value_counts()
```

## 3. Yes and no
The column *returned* says wheter the book is back. How many different values does it have? Turn it into `True` and `False` with `strip`, `lower` and `map`. Check that nothing became missing, and count `True` and `False`.

```python
# loans_raw has 10 different spellings of Yes and No
# loans only has True or False

loans["returned"] = loans["returned"].str.strip().str.lower().map(
    {"yes": True, "y": True, "no": False, "n": False})

print(loans["returned"].isna().sum())
print(loans["returned"].value_counts())

print("Old amount of unique values:", loans_raw["returned"].nunique())
print("New amount of unique values:", loans["returned"].nunique())
```
