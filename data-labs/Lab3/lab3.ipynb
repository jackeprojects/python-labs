## 2. Clean the genre
How many different values does *genre* have? It should be 6 genres. Remove spaces and capital letters with `str.strip()` and `str.lower()` and count again. Look at the values that are left with `unique()` and fix the rest with `replace`. Finish with `str.title()`. How many genres do you have, and how many loans in each?

```python
# 28 genres before cleanup
loans["genre"].unique()
```

```python
# 9 genres after first cleanup
loans["genre"] = loans["genre"].str.strip().str.lower()
loans["genre"].unique()
```

```python
# 6 genres after second cleanup
loans["genre"] = loans["genre"].replace({"children's": "children", "science fiction" : "sci-fi", "sci fi" : "sci-fi"}).str.title()
loans["genre"].unique()
```

```python
# 163 in Crime
# 135 in Children
# 90 in Fantasy
# 90 in Sci-Fi
# 86 in Romance
# 59 in History
loans["genre"].value_counts()
```

## 3. Yes and no
The column *returned* says wheter the book is back. How many different values does it have? Turn it into `True` and `False` with `strip`, `lower` and `map`. Check that nothing became missing, and count `True` and `False`.

```python
# loans_raw has 10 different spellings of Yes and No
# loans only has True or False

loans["returned"] = loans["returned"].str.strip().str.lower().map(
    {"yes": True, "y": True, "no": False, "n": False})

print(loans["returned"].isna().sum())
print(loans["returned"].value_counts())

print("Old amount of unique values:", loans_raw["returned"].nunique())
print("New amount of unique values:", loans["returned"].nunique())
```

## 4. The late fee is text
The late fee looks like *7,50 kr*. Turn it into a number. First try `astype(float)` directly and read the error. Then remove *kr*, remove the spaces, change the comma to a dot, and only then use `astype(float)`. Check the result with `describe()` and calculate the total of all fees. Try `astype(float)` on the original text first, read the error from the bottom up.

```python
# Can't convert to float since the str contains non-digits
loans_raw["late_fee"].astype(float)
```

```python
# Total fees are 7677.5 kr

loans["late_fee"] = loans["late_fee"].str.replace(
    {"kr": "", "," : "."}).str.strip().astype(float)
print(loans["late_fee"].describe())
loans["late_fee"].sum()
```

## 5. Dates
Turn *loan_date* into real dates with `pd.to_datetime(..., errors="coerce")`. How many dates became `NaT`? Find the first and the last date. Then count the loans per month with `dt.to_period("M")`, and check if the numbers add up to the whole table.

```python
# 3 dates are NaT
# First loan is 2026-01-05
# Last loan is 2026-06-27
# The monthly dates add up to 620
loans["loan_date"] = pd.to_datetime(loans["loan_date"], errors="coerce")
print(loans["loan_date"].isna().sum())
print("first date:", loans["loan_date"].min())
print("last date:", loans["loan_date"].max())
print(loans["loan_date"].dt.to_period("M").value_counts().sort_index())
```
